**Resumo**


O capítulo seis trata da limpeza e da transformação de dados, fases essenciais para uniformizar bases inconsistentes antes da análise. A seleção possibilita filtrar linhas de interesse por meio de indexação booleana, enquanto a projeção permite extrair colunas determinadas de um DataFrame. Para reorganizar o conjunto de dados, pode-se recriar os índices após a filtragem utilizando o método reset_index(). A alteração dos dados inclui operações de inserção por concatenação, bem como a atualização e remoção de registros por meio de loc(). No âmbito da formatação, o método replace() troca valores específicos, sendo especialmente útil para tratar dados ausentes ou sentinelas. A aplicação de funções com o método apply() possibilita formatar colunas inteiras de maneira eficiente usando funções personalizadas.Atributos categóricos podem igualmente ser padronizados convertendo-os em variáveis binárias por meio do método get_dummies(). Essas abordagens asseguram que algoritmos de Machine Learning funcionem adequadamente com conjuntos de dados que, originalmente, apresentavam escalas e unidades distintas. O autor ilustra a aplicação prática dessas ideias no conjunto de dados flags, produzindo uma base homogênea para classificação. Ao término, o capítulo deixa o leitor pronto para a fase de modelagem, tendo finalizado as etapas mais trabalhosas do Data Wrangling.

```python
import pandas as pd
def classe_area(area):
    if (area < 10) :
        return 'F'
    elif (area >= 10 and area < 50) :
        return 'E'
    elif (area >= 50 and area < 100) :
        return 'D'
    elif (area >= 100 and area < 500) :
        return 'C'
    elif (area >= 500 and area < 1000) :
        return 'B'
    else :
        return 'A'
flags = pd.read_csv('flags.csv')
flags['classe_area'] = flags['area'].apply(classe_area)
print(flags[['name',
'area',
'classe_area'
]])
```

```python
import pandas as pd
flags = pd.read_csv('flags.csv')
for c in flags.columns:
    if c in ['landmass', 'zone',
'language', 'religion', 'mainhue',
'topleft', 'botright']:
        dummies = pd.get_dummies(flags[c], prefix=c)
        flags = flags.join(dummies)
for c in flags.columns:
    if c in ['area', 'population',
'bars', 'stripes', 'colours',
'circles', 'crosses', 'saltires',
'quarters', 'sunstars']:
        c_max = max(flags[c])
        c_min = min(flags[c])
        flags[c] = (flags[c] - c_min) / (c_max - c_min)
flags = flags.drop(columns=['name', 'landmass',
'zone', 'language',
'religion', 'mainhue',
'topleft', 'botright',
])
print('head():'); print(flags.head())
print('-----------------------------------------------')
print('tail():'); print(flags.tail())
print('-----------------------------------------------')
flags.to_csv("flags_transf.csv", sep=",", index=False)
```

**1)**  

Uma empresa possui informações sobre funcionários e o setor em que trabalham.

A tabela possui:

nome → nome do funcionário
setor → código do setor da empresa

Os códigos são:

1 = Financeiro
2 = Tecnologia
3 = Recursos Humanos

Crie um programa em Python usando pandas para:

Criar um DataFrame com os dados abaixo:
Ana → setor 2
Carlos → setor 1
Mariana → setor 2
João → setor 3
Filtrar apenas os funcionários do setor de Tecnologia (setor == 2)

```python
import pandas as pd

dados = {
    'nome': ['Ana', 'Carlos', 'Mariana', 'João'],
    'setor': [2, 1, 2, 3]
}

funcionarios = pd.DataFrame(dados)

tecnologia = funcionarios[funcionarios['setor'] == 2]

print(tecnologia)
```

**2)** Uma empresa possui uma tabela com informações de filiais espalhadas pelo mundo.

A tabela contém:

nome → nome da filial
regiao → código da região
funcionarios → quantidade de funcionários
idioma → idioma principal
cores → quantidade de cores no logotipo

Os códigos de região são:

1 = América
2 = Europa
3 = Ásia
4 = África
5 = Oceania

Criar um DataFrame com os dados abaixo:
nome	regiao	funcionarios	idioma	cores

Alpha	5	350	Inglês	3

Beta	2	120	Francês	2

Gamma	5	180	Inglês	4

Delta	1	500	Português	5

E depois Filtre apenas as filiais:
da Oceania (regiao == 5)
com mais de 200 funcionários

```python
import pandas as pd

dados = {
    'nome': ['Alpha', 'Beta', 'Gamma', 'Delta'],
    'regiao': [5, 2, 5, 1],
    'funcionarios': [350, 120, 180, 500],
    'idioma': ['Inglês', 'Francês', 'Inglês', 'Português'],
    'cores': [3, 2, 4, 5]
}

empresas = pd.DataFrame(dados)

v = (empresas['regiao'] == 5) & (empresas['funcionarios'] > 200)

df = empresas[v]

df = df[['nome', 'cores', 'idioma', 'regiao', 'funcionarios']]

df = df.reset_index(drop=True)

print(df)
```
